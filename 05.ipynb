```python
even=0
odd=0
while 1:
    x=int(input("enter a number:"))
    if x%2==0:
        even=even+x
    else :
        odd=odd+x
    y=input("do you want to continue")
    if y=="yes":
        pass
    elif y=="no":
        break
print("sum of even",even) 
print("sum of odd",odd)
```
